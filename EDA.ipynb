```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset('titanic')
df.head()
```

```python
# Fill missing age with the median age
df['age'] = df['age'].fillna(df['age'].median())
```

```python
# Drop the deck column entirely — too many missing values (77%) to be useful
df = df.drop(columns=['deck'])
```

```python
# Drop the 2 rows with missing embarked/embark_town (negligible loss)
df = df.dropna(subset=['embarked', 'embark_town'])
```

```python
# Confirm no more missing values
df.isnull().sum()
```

```python
survival_rate = df['survived'].mean() * 100
print(f"Overall survival rate: {survival_rate:.2f}%")
df['survived'].value_counts()
```

```python
df.groupby('sex')['survived'].mean() * 100
df.groupby('sex')['survived'].value_counts()
```

```python
df.groupby('pclass')['survived'].mean() * 100
df.groupby('pclass')['survived'].value_counts()
```

```python
# Create age groups
df['age_group'] = pd.cut(df['age'], bins=[0, 12, 18, 35, 60, 80],
                          labels=['Child (0-12)', 'Teen (13-18)', 'Young Adult (19-35)', 'Adult (36-60)', 'Senior (60+)'])

df.groupby('age_group')['survived'].mean() * 100
```

```python
# Using the existing 'alone' column
df.groupby('alone')['survived'].mean() * 100
# Create family size = siblings/spouses + parents/children + self
df['family_size'] = df['sibsp'] + df['parch'] + 1

df.groupby('family_size')['survived'].mean() * 100
```
